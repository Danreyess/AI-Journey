```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI  # pyright: ignore[reportUnusedImport]
from IPython.display import Markdown, display
import ollama 
```

```python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f'OpenAI API Key exists and begins {openai_api_key[:8]}')
else:
    print('OpenAI Key not set')
```

```python
#Telling a joke
tell_a_joke = [
    {"role": "user", "content": "Tell a joke for student on the journey to becoming an expert in LLM Engineering"}
]
```

```python
client = OpenAI(api_key=openai_api_key)

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=tell_a_joke
)
```

```python
display(Markdown(response.choices[0].message.content))
```

Training vs Inference time scaling 


```python
easy_puzzle = [
    {"role": "user",
    "content": "You toss 2 coins, One of them is heads. what's the probability the other is tails?, Answer with the probability only"}
]
```

```python

response = client.chat.completions.create(
    model="gpt-5-nano",
    messages=easy_puzzle,
    reasoning_effort="minimal")

display(Markdown(response.choices[0].message.content))
```
